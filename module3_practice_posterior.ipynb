### Visualize prior, likelihood, and posterior

All three are normal densities on one axis. The posterior sits between the prior and the data, closer to whichever is more precise, and narrower than either.

```python
def npdf(x, mean, var):
    return np.exp(-(x - mean) ** 2 / (2 * var)) / np.sqrt(2 * np.pi * var)

xs = np.linspace(-4, 8, 400)
lik_var_mean = LIK_VAR / n
plt.figure(figsize=(7, 4))
plt.plot(xs, npdf(xs, PRIOR_MEAN, PRIOR_VAR), label=f"prior N({PRIOR_MEAN:.0f}, {PRIOR_VAR:.0f})")
plt.plot(xs, npdf(xs, xbar, lik_var_mean), label=f"likelihood of the mean N({xbar:.2f}, {lik_var_mean:.2f})")
plt.plot(xs, npdf(xs, seq_mean, seq_var), label=f"posterior N({seq_mean:.2f}, {seq_var:.2f})")
plt.xlabel("theta")
plt.ylabel("density")
plt.title("Prior, likelihood, and posterior")
plt.legend()
plt.tight_layout()
plt.show()
```
