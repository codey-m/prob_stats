### R-squared

R-squared is the fraction of the variance in y explained by the fitted line.

```python
pred = slope * x + intercept
ss_res = ((y - pred) ** 2).sum()
ss_tot = ((y - y.mean()) ** 2).sum()
r2 = 1 - ss_res / ss_tot
print(f"R^2 = {r2:.3f}")
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(x, y, label="data")
plt.plot(x, pred, color="red", label=f"fit: y = {intercept:.2f} + {slope:.2f}x")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Least-squares fit")
plt.legend()
plt.tight_layout()
plt.show()
```
