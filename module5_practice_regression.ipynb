### R-squared

R-squared (computed below) is the fraction of the variance in `y` explained by the fit.

```python
pred = slope * x + intercept
ss_res = ((y - pred) ** 2).sum()
ss_tot = ((y - y.mean()) ** 2).sum()
r2 = 1 - ss_res / ss_tot
print(f"R^2 = {r2:.3f}")
```
